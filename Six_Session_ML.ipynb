```python
import pandas as pd
data = {
  "Expriance":[1,2,3,5,7,8,10,12],
  "Education":[12,14,14,16,16,18,18,20],
  "Salary":[35,40,45,55,65,72,85,100]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
import seaborn as sms
import matplotlib.pyplot as plt

sms.pairplot(df)
plt.show()
```

```python
X = df[["Expriance", "Education"]]
y = df["Salary"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2) # test data 20%, traning data 80%
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.intercept_
```

```python
model.coef_
```

```python
model.predict([[10, 18]])
```
