```python
import pandas as pd
data = {"Exprience": [1,2,3,4,5,6], "Salary": [35,40,48,55,62,70]}
df = pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["Exprience"], df["Salary"])
plt.xlabel("Exprience")
plt.ylabel("Salary")
plt.title("Exprince Vs Salary")

plt.show()
```

```python
import pandas as pd
data = {"AdvertiesmentSpend": [10,20,30,40,50,60], "Sales": [350,400,480,550,620,700]}
df = pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["AdvertiesmentSpend"], df["Sales"])
plt.xlabel("AdvertiesmentSpend")
plt.ylabel("Sales")
plt.title("AdvertiesmentSpend Vs Sales")

plt.show()
```

```python
prediction = 75
actual = 70
residual = actual - prediction
print(residual)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X = df[["AdvertiesmentSpend"]]
y = df["Sales"]
```

```python
model = LinearRegression()
```

```python
model
```

```python
model.fit(X,y)
```

```python
model.intercept_
```

```python
model.coef_[0]
```

```python
model.predict([[105]])
```

```python
model.predict([[105]])
```

```python
model.predict([[200]])
```

```python
model.predict([[105],[63],[70]])
```

```python
prediction = model.predict(X)
plt.plot(X,prediction,color ="red")
plt.scatter(df["AdvertiesmentSpend"], df["Sales"])
plt.xlabel("AdvertiesmentSpend")
plt.ylabel("Sales")
plt.title("AdvertiesmentSpend Vs Sales")

plt.show()
```
