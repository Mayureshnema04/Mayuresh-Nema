```python
def calculate_gst(price,gst_persentage):
  gst = price * gst_persentage / 100
  return gst
```

```python
def calculate_delivery_charges(price,delivery_charges):
  delivery_charges = price * delivery_charges / 100
  return delivery_charges
```

```python
price = 1000
discount_persentage = 10
delivery_charges = 5
gst_persentage = 5
```

```python
final_price = price - calculate_discount(price,discount_persentage) + calculate_delivery_charges(price,delivery_charges) + calculate_gst(price,gst_persentage)
```
