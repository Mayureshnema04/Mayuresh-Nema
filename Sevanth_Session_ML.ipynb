```python
#Logitstic Regression
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
```

```python
# Studying less than 3 hours chance are to fail
X = np.array([[1],[2],[3],[4],[5],[6],[7]])
y = np.array([0,0,0,1,1,1,1])
```

```python
model = LogisticRegression()
```

```python
model.fit(X,y)
```

```python
model.predict([[5]])
```

```python
model.predict([[10]])
```

```python
model.predict_proba([[5]])
```

```python
model.predict_proba([[1]])
```

```python
model.predict_proba([[2]])
```

```python
X_test = np.array([[2],[3],[4],[6],[8]])
```

```python
model.predict(X_test)
```

```python
model.predict_proba(X_test)
```

```python
model.predict_proba([[0]])
```

```python
def predict_pass(hours):
  prediction = model.predict([[hours]])

  return prediction
```

```python
predict_pass(5)
```

```python
predict_pass(5)[0]
```

```python
if predict_pass(5)[0] == 1:
  print("Student has passed the exam")
else:
    print("Student has failed the exam")
```
